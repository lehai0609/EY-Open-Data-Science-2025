## 2. Building footprints

***a. Buffer Creation:***

Create a circular buffer (e.g., 100 meters) around each UHI point. This defines the area we'll use to aggregate nearby building features.

***b. Spatial Join:***

Use a spatial join to determine which building footprints fall within each UHI point's buffer. Then, you can calculate summary statistics (count, mean height, total area, etc.) for those buildings.

```python
# Create a 100-meter buffer around each UHI point
df_uhi['buffer'] = df_uhi.geometry.buffer(100)

# Perform a spatial join to find buildings that intersect with each buffer
joined = gpd.sjoin(gdf_buildings, df_uhi.set_geometry('buffer'), how='inner', predicate='intersects')

# Aggregate the building data by each UHI point (index_right)
agg_buildings = joined.groupby('index_right').agg(
    building_count=('heightroof', 'count'),
    mean_height=('heightroof', 'mean'),
    total_area=('calculated_area_sqm', 'sum')
).reset_index()

# Merge the aggregated building features back to the UHI dataframe
df_uhi = df_uhi.reset_index().merge(agg_buildings, left_index=True, right_on='index_right', how='left')

# Assuming a 100m buffer: area of buffer = π * (100^2) ≈ 31416 m²
buffer_area = 31416  # square meters
df_uhi['building_area_ratio'] = df_uhi['total_area'] / buffer_area
```

## 3. NDVI Aggregation

***a. Spatial Join***

Join the NDVI GeoDataFrame (which contains pixel-level NDVI values) with the UHI buffers (100m).

***b. Aggregation:***

Group by each UHI point (using the corresponding index from the UHI data) and compute the mean (and optionally other statistics) of NDVI values within that buffer.

***c. Merge back:***

Merge these aggregated NDVI features back into the UHI dataset.

```python
# Spatial join: find NDVI pixels that intersect with each UHI point buffer.
joined_ndvi = gpd.sjoin(gdf_ndvi, df_uhi.set_geometry('buffer'),
                        how='inner', predicate='intersects')

# Aggregate NDVI values by UHI point (index_right)
agg_ndvi = joined_ndvi.groupby('index_right').agg(
    mean_ndvi=('NDVI', 'mean'),
    median_ndvi=('NDVI', 'median'),
    std_ndvi=('NDVI', 'std')
).reset_index()

# Merge aggregated NDVI features back to the UHI dataframe
df_uhi = df_uhi.reset_index().merge(agg_ndvi, left_index=True, right_on='index_right', how='left')
```

## 4. LST Aggregation

***a. Spatial Join***

Join the LST GeoDataFrame with the UHI buffers.

***b. Aggregation:***

Group by UHI point and compute the mean, maximum, or other relevant statistics of the LST values.

***c. Merge back:***

Merge the aggregated LST data back into the UHI dataset.

```python
df_uhi = df_uhi.drop(columns=['level_0', 'index_right_x', 'index_right_y'])
# Spatial join: find LST pixels that intersect with each UHI point buffer.
joined_lst = gpd.sjoin(gdf_lst, df_uhi.set_geometry('buffer'),
                       how='inner', predicate='intersects')

# Aggregate LST values by UHI point (index_right)
agg_lst = joined_lst.groupby('index_right').agg(
    mean_lst=('LST', 'mean'),
    max_lst=('LST', 'max')
).reset_index()

# Merge aggregated LST features back to the UHI dataframe
df_uhi = df_uhi.reset_index().merge(agg_lst, left_index=True, right_on='index_right', how='left')
```

## 5. Weather data Integration

***a. Temporal matching***

For each UHI point (which has a timestamp), select weather observations within a time window around that measurement (e.g., ±15 minutes).

***b. Aggregation:***

Compute summary statistics (mean, median) for weather variables within that time window.

***c. Merge back:***

Merge the aggregated LST data back into the UHI dataset.

```python
# Convert weather 'Date / Time' to datetime if not already done
df_weather['Date / Time'] = pd.to_datetime(df_weather['Date / Time'], errors='coerce')

# Define a time window (e.g., ±15 minutes) for matching
time_window = pd.Timedelta('15 minutes')

def aggregate_weather(uhi_time):
    # Filter weather records close to the UHI timestamp
    weather_subset = df_weather[(df_weather['Date / Time'] >= uhi_time - time_window) &
                                (df_weather['Date / Time'] <= uhi_time + time_window)]
    if len(weather_subset) == 0:
        return pd.Series({
            'mean_air_temp': None,
            'mean_humidity': None,
            'mean_wind_speed': None,
            'mean_solar_flux': None
        })
    return pd.Series({
        'mean_air_temp': weather_subset['Air Temp at Surface [degC]'].mean(),
        'mean_humidity': weather_subset['Relative Humidity [percent]'].mean(),
        'mean_wind_speed': weather_subset['Avg Wind Speed [m/s]'].mean(),
        'mean_solar_flux': weather_subset['Solar Flux [W/m^2]'].mean()
    })

# Apply the aggregation function to each UHI point
weather_features = df_uhi['datetime'].apply(aggregate_weather)

# Merge the weather features into the UHI dataframe
df_uhi = pd.concat([df_uhi, weather_features], axis=1)
```
