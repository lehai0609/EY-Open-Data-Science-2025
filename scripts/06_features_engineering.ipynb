```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
import matplotlib.pyplot as plt
```

```python
df_uhi = gpd.read_parquet("../output/uhi_data_processed.parquet")
gdf_buildings = gpd.read_parquet("../output/building_data_processed.parquet")
gdf_ndvi = gpd.read_parquet("../output/sentinel2.parquet")
gdf_lst = gpd.read_parquet("../output/landsat_lst.parquet")
df_weather = pd.read_parquet("../output/weather_data_processed.parquet")
```

```python
# Define a buffer radius in meters (e.g., 100m)
buffer_radius = 100

# Create a buffer around each UHI point
df_uhi['buffer'] = df_uhi.geometry.buffer(buffer_radius)

# Spatial join: Count buildings within each UHI buffer
# Ensure building footprints (gdf_buildings) are in EPSG:2263
buildings_within = gpd.sjoin(gdf_buildings, df_uhi[['buffer', 'geometry']], how='left', predicate='intersects')
# Count the number of buildings per UHI point
building_counts = buildings_within.groupby('index_right').size().rename('building_count')
# Merge back to UHI DataFrame
df_uhi = df_uhi.join(building_counts, how='left')
df_uhi['building_count'] = df_uhi['building_count'].fillna(0)

# For NDVI: Use a nearest neighbor approach
gdf_uhi = gpd.GeoDataFrame(df_uhi.copy(), geometry='geometry', crs="EPSG:2263")

# Spatial join NDVI - find nearest NDVI point for each UHI point
gdf_uhi = gdf_uhi.sjoin_nearest(gdf_ndvi[['NDVI', 'geometry']], how='left', distance_col='ndvi_dist')

# Similarly for LST
gdf_uhi = gdf_uhi.sjoin_nearest(gdf_lst[['LST', 'geometry']], how='left', distance_col='lst_dist', rsuffix='_lst')

# For weather, we can assign each UHI point the nearest weather station measurement.
# First, create GeoDataFrame for weather points. You may need station coordinates; here, assume you have them.
# For this example, we use a simple method by assuming weather points are available in df_weather with a 'geometry' column.
# If they are not, you must create them using known station coordinates.

# For demonstration, let's convert the weather DataFrame to a GeoDataFrame:
# (Assuming you have a lookup for station coordinates; here, we use approximate centroids for Bronx and Manhattan)
station_coords = {'Bronx': (981000, 260000), 'Manhattan': (980000, 260000)}  # adjust as needed

df_weather['geometry'] = df_weather['location'].apply(lambda loc: Point(station_coords[loc]))
gdf_weather = gpd.GeoDataFrame(df_weather, geometry='geometry', crs="EPSG:2263")

# Spatial join: assign nearest weather station to each UHI point
gdf_uhi = gdf_uhi.sjoin_nearest(gdf_weather, how='left', distance_col='weather_dist')
```
